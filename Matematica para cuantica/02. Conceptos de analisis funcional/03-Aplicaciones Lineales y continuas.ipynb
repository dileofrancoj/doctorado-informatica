# Aplicaciones Lineales y continuas

Sean $X$ e $Y$ espacios vectoriales sobre un cuerpo $\mathbb{K}$ recordemos que una aplicacion $T:X \to Y$ es lineal si:

1. $T(x+y) = T(x) + T(y)$ (Linealidad)
2. $T(\alpha x) = \alpha T(x)$ (Homogeneidad)
  
Con esto en mente, nos interesa saber cuando una aplicacion lineal es continua por lo que se dará una caracterizacion de la continuidad de una aplicacion lineal entre espacios normados.
