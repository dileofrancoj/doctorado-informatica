# Introducción

La motivación más simple para el concepto de norma sobre un $\mathbb{K}$ espacio vectorial no es otro más que la generalización del concepto de valor absoluto y módulo en $\mathbb{R}$ y $\mathbb{C}$ respectivamente.

## Definición

Sea $X$ un espacio vectorial sobre $\mathbb{K}$. Una *norma* sobre X es una aplicación $x \to ||x||$ de $X$ en $\mathbb{R}$ que verifica:

1. $||x|| \geq 0 $ con $||x|| = 0 \iff x = 0$ (Positiva o cero)
2. $||x+y|| \leq ||x|| + ||y||$, $\forall x,y \in X$. Desigualdad triangular.
3. $||\alpha x|| = |\alpha| ||x||$ Homogeneidad.

A partir de esto, un *espacio normado* es un par $(X, || \cdot\ ||)$ donde $X$ es un $\mathbb{K}$ espacio vectorial y $|| \cdot\ ||$ es una norma sobre $X$.

Con esto en mente, la aplicacion norma, nos permite medir vectores. Además, toda norma induce una función distancia $d(x,y):= || x-y ||$. En este notebook veremos que en un espacio vectorial podemos definir infinitas normas.

### Ejemplos: Demostraremos de que $L^1$, $L^2$ y $L^\infty$ son normas

Sea $ x=(x_1,x_2,\ldots,x_n)\in\mathbb R^n$


### Norma $L^1$

Definimos:

$$
\boxed{
\|x\|_1
=
\sum_{i=1}^{n}|x_i|
}
$$

---

#### 1. Positividad

Como $|x_i|\geq0$ $\forall i$ por definición de valor absoluto, entonces:

$$
\sum_{i=1}^{n}|x_i|\geq0.
$$

Además, $\sum_{i=1}^{n}|x_i|=0$ solamente ocurre cuando: $|x_i|=0$ para todo $i$.

Por lo tanto: $x_i=0$ para todo $i$.

Luego:

$$
\|x\|_1=0
\iff
x=0.
$$


---

#### 2. Homogeneidad

$$
\begin{aligned}
\|\alpha x\|_1
&=
\sum_{i=1}^{n}|\alpha x_i|
\\
&=
\sum_{i=1}^{n}|\alpha||x_i|
\\
&=
|\alpha|
\sum_{i=1}^{n}|x_i|
\\
&=
|\alpha|\|x\|_1.
\end{aligned}
$$


---

#### 3. Desigualdad triangular

$$
\begin{aligned}
\|x+y\|_1
&=
\sum_{i=1}^{n}|x_i+y_i|
\\
&\leq
\sum_{i=1}^{n}
(|x_i|+|y_i|)
\\
&=
\sum_{i=1}^{n}|x_i|
+
\sum_{i=1}^{n}|y_i|
\\
&=
\|x\|_1+\|y\|_1.
\end{aligned}
$$

Por lo tanto:

$$
\boxed{
\|\cdot\|_1
\text{ es una norma.}
}
$$

### Norma $L^2$ (norma euclídea usual)

Definimos:

$$
\boxed{
\|x\|_2
=
\sqrt{
\sum_{i=1}^{n}x_i^2
}
}
$$


#### 1. Positividad

Como

$$
x_i^2\geq0,
$$

tenemos:

$$
\sum_{i=1}^{n}x_i^2\geq0.
$$

Luego:

$$
\|x\|_2\geq0.
$$

Además,

$$
\|x\|_2=0
$$

implica:

$$
\sum_{i=1}^{n}x_i^2=0.
$$

Esto solamente ocurre cuando:

$$
x_i=0 \forall i
$$



Por lo tanto:

$$
\|x\|_2=0
\iff
x=0.
$$

#### 2. Homogeneidad

$$
\begin{aligned}
\|\alpha x\|_2
&=
\sqrt{
\sum_{i=1}^{n}(\alpha x_i)^2
}
\\
&=
\sqrt{
\alpha^2
\sum_{i=1}^{n}x_i^2
}
\\
&=
|\alpha|
\sqrt{
\sum_{i=1}^{n}x_i^2
}
\\
&=
|\alpha|\|x\|_2.
\end{aligned}
$$


#### 3. Desigualdad triangular

La desigualdad triangular corresponde a la desigualdad de Minkowski:

$$
\sqrt{
\sum_i(x_i+y_i)^2
}
\leq
\sqrt{\sum_i x_i^2}
+
\sqrt{\sum_i y_i^2}.
$$

Por lo tanto:

$$
\boxed{
\|x+y\|_2
\leq
\|x\|_2+\|y\|_2
}
$$


Luego:

$$
\boxed{
\|\cdot\|_2
\text{ es una norma.}
}
$$

```python
import numpy as np
import matplotlib.pyplot as plt


x = np.linspace(-1.5, 1.5, 500)
y = np.linspace(-1.5, 1.5, 500)

X, Y = np.meshgrid(x, y)



L1 = np.abs(X) + np.abs(Y)

L2 = np.sqrt(X**2 + Y**2)

Linf = np.maximum(np.abs(X), np.abs(Y))



plt.figure(figsize=(8,8))



plt.contour(
    X,
    Y,
    L1,
    levels=[1],
    linewidths=3,
    label="L1"
)

plt.contour(
    X,
    Y,
    L2,
    levels=[1],
    linewidths=3,
    label="L2"
)

plt.contour(
    X,
    Y,
    Linf,
    levels=[1],
    linewidths=3,
    label="L∞"
)


plt.text(
    0.75,
    0.35,
    "$L^1$",
    fontsize=14
)

plt.text(
    0.75,
    0.75,
    "$L^2$",
    fontsize=14
)

plt.text(
    1.05,
    0.0,
    "$L^\\infty$",
    fontsize=14
)


# Ejes

plt.axhline(0)
plt.axvline(0)

plt.xlabel("$x_1$")
plt.ylabel("$x_2$")


plt.title(
    "Bolas unitarias en $\\mathbb{R}^2$ según distintas normas"
)


plt.grid(True)

plt.axis("equal")

plt.xlim(-1.5,1.5)
plt.ylim(-1.5,1.5)

plt.show()
```
