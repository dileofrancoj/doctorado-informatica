# Introducción

Como mencionamos, una de las consecuencias del hecho de que un espacio vectorial $X$ esté dotado de una norma $||\cdot||$ es que esta induce una distancia $d(x,y)$ en X. Concretamente se define la aplicacion:

$$
d: X \times X \to \mathbb{R},
$$

tal que $d(x,y) := ||x-y|| = ||y-x||$.

Esta distancia inducida nos permite

1. Definir bolas abiertas, cerradas, bolas unidad y esfera unidad.
2. Hablar de convergencia de sucesiones
3. Hablar de completitud
4. etc.

En virtud de ser conciso voy a profundizar levemente sobre el punto 2 y 3.

## 2. Convergencia de sucesiones

Dada una sucesión $\{ x_n\}$ de elementos de X se dice que es convergente si:

$\exists x \in X: \forall \epsilon \gt 0, \exists n_0 \in \mathbb{N} / n \geq n_0 \implies ||x_n - x|| < \epsilon$

Es decir, hay un elemento $x$ de $X$ y existe un natural $(n_0)$ a partir del cual la distancia entre $x_n$ y $x$ sea tan pequeña como se desee. 

Dado que ese valor $x$ es único se dice que la sucesión $\{ x_n \} $ converge a $x$ o que $x$ es el *limite* de la sucesión.

## 3. Completitud

### Sucesión de Cauchy

Se dice que una sucesión $\{ x_n \}$ de números reales es una sucesión de Cauchy si $\forall \epsilon \gt 0$ existe un número natural $k \in \mathbb{N}$ tal que para todos los números naturales $n,m \geq k$ se verifica que: $||x_n - x_m|| < \epsilon$.

Notar la diferencia respecto a la definción de convergencia que dimos antes.

Dado un valor positivo arbitrario ($\epsilon$), existe un momento a partir del cual la distancia entre dos términos *cualesquiera* de la sucesión es menor al valor arbitrario establecido.

### Ejemplo

La sucesión $\{\frac{1}{n}\}$ es una sucesión de Cauchy.

Sea $\epsilon \gt 0$ tomaremos un $k \gt \frac{2}{\epsilon}$. Si $n,m \gt k$ entonces:

$\frac{1}{n} < \frac{1}{k} < \frac{\epsilon}{2}$ y $\frac{1}{m} < \frac{\epsilon}{2}$

$|\frac{1}{n} - \frac{1}{m}| \leq \frac{1}{n} + \frac{1}{m} < \frac{\epsilon}{2} + \frac{\epsilon}{2} = \epsilon$ 

$\therefore |\frac{1}{n} - \frac{1}{m}| < \epsilon$ es de Cauchy.

## Importante

Toda sucesión de Cauchy es una sucesión convergente y esta propiedad hace que el cuerpo en el que se está trabajando $\mathbb{K}$ que puede ser $\mathbb{R}$ o $\mathbb{C}$ sea completo. Esto, informalmente quiere decir que no hay huecos. En el caso de $\mathbb{R}$ que estamos trabajando (la recta numérica) quiere decir que a cada punto de la recta le corresponde un número real.

## Toda sucesión Convergente es de Cauchy

Sea $\epsilon \gt 0.$ Dado que $\{x_n \}$ es convergente por hipótesis, supongamos que converge a $L$. Tomemos un número real arbitrario, digamos $\frac{\epsilon}{2}$. Como es convergente existe un $n_0 \in \mathbb{N}$ tal que si $n \geq n_0 \implies ||x_n - L || < \frac{\epsilon}{2}$

Sea $k=n_0$. Si $n,m \geq k$ entonces:

$||x_n - x_m|| = ||x_n - L + L -x_m|| <= ||x_n -L|| + ||L-x_m|| = || x_n - L|| + ||x_m -L|| < \frac{\epsilon}{2} + \frac{\epsilon}{2} = \epsilon$

$\therefore ||x_n - x_m|| < \epsilon$

Por lo que $\{ x_n\} $ es de Cauchy.

## Conclusiones

Como dijimos, toda norma $|| \cdot || $ induce una métrica. Un espacio métrico $(X,d)$ es completo si toda sucesión de Cauchy contenida en X converge a un elemento de X.

La idea intuitiva, como hablamos es que no hay nada "pegado" a X que no esté en $(X,d)$.

Ejemplos de esto:

1. El conjunto de números reales $\mathbb{R}$ es completo con la métrica inducida por la norma.
2. $\mathbb{R} - {0}$ no es completo dado que $\{x_n\} = \{\frac{1}{n}\}$ es de Cauchy ya que el cero está excluiddo del conjunto.
3. Todo espacio vectorial normado de dimensión finita es completo si está definido sobre un cuerpo completo.
