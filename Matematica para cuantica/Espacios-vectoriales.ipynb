# Introducción

Lo primero que repasaremos en esta travesía es la definición de **espacio vectorial** y, en consecuencia, la de **vector**. Dominar estos conceptos será fundamental, ya que tanto la mecánica cuántica como, en particular, la computación cuántica, se formulan sobre espacios vectoriales con estructura adicional llamados **espacios de Hilbert**.

En este contexto, los estados cuánticos se representan mediante vectores de un espacio de Hilbert, mientras que las magnitudes físicas observables se describen mediante operadores hermíticos.

El gran salto conceptual consiste en comprender que la idea de **vector** es mucho más amplia que la de una flecha en el plano o en el espacio. Un vector es simplemente un elemento de un espacio vectorial. Por lo tanto, siempre que un conjunto, junto con las operaciones de suma y producto por escalares, satisfaga los diez axiomas que veremos a continuación, sus elementos serán considerados vectores.

La flecha que solemos dibujar en el pizarrón para representar un elemento de $\mathbb{R}^2$ o $\mathbb{R}^3$ no es más que un caso particular. Desde esta perspectiva, sucesiones, polinomios, matrices, funciones (bajo las operaciones adecuadas) e incluso objetos mucho más abstractos pueden desempeñar el papel de vectores. Esta generalización es uno de los pilares del álgebra lineal moderna y será la base matemática sobre la que construiremos la teoría de la computación cuántica.


# De los espacios vectoriales a los espacios normados

Durante estas prácticas los espacios vectoriales estarán definidos sobre el cuerpo de los números reales $\mathbb{R}$ o sobre $\mathbb{C}$ Y $\mathbb{K}$ denotará a ambos cuerpos de forma indistinta.

# Espacios vectoriales de dimensión arbitraria

Recordemos que un $\mathbb{K}$ espacio vectorial es un conjunto $X \neq \emptyset$ dotado con dos operaciones:

$+: X \times X \to X  : (x,y) \to x+y$

$* : \mathbb{K} \times X \to X  : (a,x) \to ax$

Tales que $\forall x,y,z \in X$ y $\forall \alpha, \beta \in \mathbb{K}$ se verifican los siguientes axiomas:

(1) $(x+y) + z = x + (y+z)$ Asociatividad de la operación +.

(2) $ x + y = y + x$ Conmutatividad respecto a +

(3) $ \exists! e \in X: e + x = x + e = x $ Neutro de la suma.

(4) $ \forall x \in X \exists -x \in X : x + (-x) = e$ Inverso aditivo de la operación suma.

(5) $ \alpha (x+y) = \alpha x + \alpha y$ Distributividad del producto respecto a la suma.

(6) $ (\alpha + \beta)x = \alpha x + \beta x$ Distributividad de la suma de escalares respecto al producto de vectores.

(7) $ (\alpha \beta)x = \alpha(\beta x)$ Asociatividad del producto.

(8) $\exists! e' \in \mathbb{K} : e' x = x e' = x$

Además la suma de elementos de $X$ debe ser cerrada, es decir $x+y \in X$ y también debe serlo el producto de un elemento de X por un escalar: $\alpha x \in X$.

$\textbf{Los elementos de un espacio vectorial X se denomina vectores mientras que los elementos del cuerpo $\mathbb{K}$ se llaman escalares.}$

## Ejemplos

1. $ \mathbb{R}^n$ es un espacio vectorial respecto a la suma y producto usuales de vectores.
2. $ \mathbb{C}^n$ es un espacio vectorial respecto a la suma y producto usuales de vectores.
3. El conjunto de sucesiones finitas es un espacio vectorial respecto a la suma y producto usual.
4. El conjunto de funciones continuas en un intervalo cerrado y acotado es un espacio vectorial respecto a la suma y producto de funciones.
5. El conjunto de matrices $M_{m\times n}(\mathbb{R})$ forman un espacio vectorial respecto a la suma de matrices y el producto por escalar.

### Demostración de que $\mathbb{R}^n$ es un espacio vectorial

Sea

$$
\mathbb{R}^n=
\left\{
(x_1,\ldots,x_n):x_i\in\mathbb R
\right\},
$$

con las operaciones

$$
(x_1,\ldots,x_n)+(y_1,\ldots,y_n)
=
(x_1+y_1,\ldots,x_n+y_n),
$$

y

$$
\lambda(x_1,\ldots,x_n)
=
(\lambda x_1,\ldots,\lambda x_n),
\qquad
\lambda\in\mathbb R.
$$

Sean

$$
u=(u_1,\ldots,u_n),
\qquad
v=(v_1,\ldots,v_n),
\qquad
w=(w_1,\ldots,w_n),
$$

y sean

$$
\alpha,\beta\in\mathbb R.
$$

Debemos verificar los diez axiomas de espacio vectorial.

---

#### 1. Clausura de la suma

Como

$$
u_i,v_i\in\mathbb R
$$

y $\mathbb R$ es cerrado bajo la suma,

$$
u_i+v_i\in\mathbb R
$$

para todo $i$.

Por lo tanto,

$$
u+v
=
(u_1+v_1,\ldots,u_n+v_n)
\in\mathbb R^n.
$$

---

#### 2. Conmutatividad de la suma

$$
\begin{aligned}
u+v
&=(u_1+v_1,\ldots,u_n+v_n)\\
&=(v_1+u_1,\ldots,v_n+u_n)\\
&=v+u,
\end{aligned}
$$

porque la suma en $\mathbb R$ es conmutativa.

---

#### 3. Asociatividad de la suma

$$
\begin{aligned}
(u+v)+w
&=((u_1+v_1)+w_1,\ldots,(u_n+v_n)+w_n)\\
&=(u_1+(v_1+w_1),\ldots,u_n+(v_n+w_n))\\
&=u+(v+w),
\end{aligned}
$$

ya que la suma en $\mathbb R$ es asociativa.

---

#### 4. Existencia del elemento neutro

Definimos

$$
0=(0,\ldots,0).
$$

Entonces,

$$
u+0
=
(u_1+0,\ldots,u_n+0)
=
u.
$$

---

#### 5. Existencia del inverso aditivo

Sea

$$
-u=(-u_1,\ldots,-u_n).
$$

Entonces,

$$
u+(-u)
=
(u_1-u_1,\ldots,u_n-u_n)
=
(0,\ldots,0).
$$

---

#### 6. Clausura del producto por escalares

Como

$$
\alpha,u_i\in\mathbb R,
$$

se tiene

$$
\alpha u_i\in\mathbb R.
$$

Luego,

$$
\alpha u
=
(\alpha u_1,\ldots,\alpha u_n)
\in\mathbb R^n.
$$

---

#### 7. Distributividad respecto de la suma de vectores

$$
\begin{aligned}
\alpha(u+v)
&=(\alpha(u_1+v_1),\ldots,\alpha(u_n+v_n))\\
&=(\alpha u_1+\alpha v_1,\ldots,\alpha u_n+\alpha v_n)\\
&=\alpha u+\alpha v.
\end{aligned}
$$

---

#### 8. Distributividad respecto de la suma de escalares

$$
\begin{aligned}
(\alpha+\beta)u
&=((\alpha+\beta)u_1,\ldots,(\alpha+\beta)u_n)\\
&=(\alpha u_1+\beta u_1,\ldots,\alpha u_n+\beta u_n)\\
&=\alpha u+\beta u.
\end{aligned}
$$

---

#### 9. Compatibilidad del producto por escalares

$$
\begin{aligned}
\alpha(\beta u)
&=\alpha(\beta u_1,\ldots,\beta u_n)\\
&=(\alpha\beta u_1,\ldots,\alpha\beta u_n)\\
&=(\alpha\beta)u.
\end{aligned}
$$

---

#### 10. Existencia del elemento identidad

$$
1u
=
(1u_1,\ldots,1u_n)
=
u.
$$

---

Como se verifican todos los axiomas de espacio vectorial, concluimos que

$$
(\mathbb R^n,+,\cdot)
$$

es un espacio vectorial sobre el cuerpo $\mathbb R$.

---

### Caso de $\mathbb{C}^n$

La demostración es exactamente la misma. Basta reemplazar

$$
\mathbb R
\longrightarrow
\mathbb C.
$$

En efecto,

- los vectores son

$$
(z_1,\ldots,z_n),
\qquad
z_i\in\mathbb C,
$$

- los escalares pertenecen a $\mathbb C$,

- todas las propiedades utilizadas (conmutatividad, asociatividad, distributividad, existencia de neutro e inversos) siguen siendo válidas porque $\mathbb C$ también es un cuerpo.

Por lo tanto,

$$
(\mathbb C^n,+,\cdot)
$$

es un espacio vectorial sobre $\mathbb C$.

### El espacio de las sucesiones finitas es un espacio vectorial

Sea

$$
\mathcal{F}=
\left\{
(a_1,a_2,\ldots):
a_i\in\mathbb{R}
\text{ y existe }N\in\mathbb{N}\text{ tal que }a_n=0\text{ para todo }n>N
\right\}.
$$

Es decir, una sucesión pertenece a $\mathcal{F}$ si solamente tiene una cantidad finita de términos distintos de cero.

Definimos las operaciones

$$
(a_n)+(b_n)=(a_n+b_n),
$$

y

$$
\lambda(a_n)=(\lambda a_n),
\qquad
\lambda\in\mathbb R.
$$

Sean

$$
u=(u_n),\qquad
v=(v_n),\qquad
w=(w_n),
$$

y

$$
\alpha,\beta\in\mathbb R.
$$

Verificaremos los axiomas de espacio vectorial.

---

#### 1. Clausura de la suma

Como $u$ y $v$ poseen solamente una cantidad finita de términos distintos de cero, existen enteros $N_u$ y $N_v$ tales que

$$
u_n=0
\quad\text{para }n>N_u,
$$

y

$$
v_n=0
\quad\text{para }n>N_v.
$$

Sea

$$
N=\max\{N_u,N_v\}.
$$

Entonces,

$$
u_n+v_n=0
\qquad
\text{para todo }n>N.
$$

Por lo tanto,

$$
u+v\in\mathcal F.
$$

---

#### 2. Conmutatividad de la suma

Como la suma de números reales es conmutativa,

$$
u_n+v_n=v_n+u_n,
$$

para todo $n$.

Luego,

$$
u+v=v+u.
$$

---

#### 3. Asociatividad de la suma

Para cada componente,

$$
(u_n+v_n)+w_n
=
u_n+(v_n+w_n),
$$

porque la suma en $\mathbb R$ es asociativa.

Así,

$$
(u+v)+w=u+(v+w).
$$

---

#### 4. Existencia del elemento neutro

La sucesión

$$
0=(0,0,0,\ldots, 0)
$$

pertenece a $\mathcal F$.

Además,

$$
u+0=u.
$$

---

#### 5. Existencia del inverso aditivo

Definimos

$$
-u=(-u_n).
$$

Como cambiar el signo no modifica la cantidad de términos distintos de cero,

$$
-u\in\mathcal F.
$$

Además,

$$
u+(-u)=0.
$$

---

#### 6. Clausura del producto por escalares

Sea

$$
\alpha\in\mathbb R.
$$

Si

$$
u_n=0
\quad
\text{para }n>N,
$$

entonces

$$
\alpha u_n=0
\quad
\text{para }n>N.
$$

Por lo tanto,

$$
\alpha u\in\mathcal F.
$$

---

#### 7. Distributividad respecto de la suma de vectores

Para cada componente,

$$
\alpha(u_n+v_n)
=
\alpha u_n+\alpha v_n.
$$

Por consiguiente,

$$
\alpha(u+v)
=
\alpha u+\alpha v.
$$

---

#### 8. Distributividad respecto de la suma de escalares

Para cada componente,

$$
(\alpha+\beta)u_n
=
\alpha u_n+\beta u_n.
$$

Luego,

$$
(\alpha+\beta)u
=
\alpha u+\beta u.
$$

---

#### 9. Compatibilidad del producto por escalares

Para todo $n$,

$$
\alpha(\beta u_n)
=
(\alpha\beta)u_n.
$$

Así,

$$
\alpha(\beta u)
=
(\alpha\beta)u.
$$

---

#### 10. Existencia del elemento identidad

Como

$$
1u_n=u_n,
$$

se obtiene

$$
1u=u.
$$

---

Como todos los axiomas se verifican, concluimos que

$$
\boxed{\mathcal F\text{ es un espacio vectorial sobre }\mathbb R.}
$$

### El conjunto de las funciones continuas es un espacio vectorial

Sea

$$
C([a,b])
=
\left\{
f:[a,b]\rightarrow\mathbb R
\;:\;
f\text{ es continua}
\right\}.
$$

Definimos

$$
(f+g)(x)
=
f(x)+g(x),
$$

y

$$
(\lambda f)(x)
=
\lambda f(x),
\qquad
\lambda\in\mathbb R.
$$

Sean

$$
f,g,h\in C([a,b]),
$$

y

$$
\alpha,\beta\in\mathbb R.
$$

Verificaremos los axiomas.

---

#### 1. Clausura de la suma

Como la suma de funciones continuas es continua,

$$
f+g\in C([a,b]).
$$

---

#### 2. Conmutatividad de la suma

Para todo $x$,

$$
(f+g)(x)
=
f(x)+g(x)
=
g(x)+f(x)
=
(g+f)(x).
$$

Luego,

$$
f+g=g+f.
$$

---

#### 3. Asociatividad de la suma

Para todo $x$,

$$
((f+g)+h)(x)
=
(f(x)+g(x))+h(x)
=
f(x)+(g(x)+h(x))
=
(f+(g+h))(x).
$$

Por lo tanto,

$$
(f+g)+h=f+(g+h).
$$

---

#### 4. Existencia del elemento neutro

La función

$$
0(x)=0
$$

es continua.

Además,

$$
(f+0)(x)
=
f(x)+0
=
f(x).
$$

Por lo tanto,

$$
f+0=f.
$$

---

#### 5. Existencia del inverso aditivo

Definimos

$$
(-f)(x)
=
-f(x).
$$

Como el producto de una función continua por el escalar $-1$ sigue siendo continuo,

$$
-f\in C([a,b]).
$$

Además,

$$
f+(-f)=0.
$$

---

#### 6. Clausura del producto por escalares

Como

$$
(\alpha f)(x)
=
\alpha f(x),
$$

y el producto de una función continua por un escalar continúa siendo continuo,

$$
\alpha f\in C([a,b]).
$$

---

#### 7. Distributividad respecto de la suma de funciones

Para todo $x$,

$$
\begin{aligned}
(\alpha(f+g))(x)
&=\alpha(f(x)+g(x))\\
&=\alpha f(x)+\alpha g(x)\\
&=(\alpha f+\alpha g)(x).
\end{aligned}
$$

Luego,

$$
\alpha(f+g)
=
\alpha f+\alpha g.
$$

---

#### 8. Distributividad respecto de la suma de escalares

Para todo $x$,

$$
\begin{aligned}
((\alpha+\beta)f)(x)
&=(\alpha+\beta)f(x)\\
&=\alpha f(x)+\beta f(x)\\
&=(\alpha f+\beta f)(x).
\end{aligned}
$$

Así,

$$
(\alpha+\beta)f
=
\alpha f+\beta f.
$$

---

#### 9. Compatibilidad del producto por escalares

Para todo $x$,

$$
\alpha(\beta f)(x)
=
\alpha(\beta f(x))
=
(\alpha\beta)f(x).
$$

Por lo tanto,

$$
\alpha(\beta f)
=
(\alpha\beta)f.
$$

---

#### 10. Existencia del elemento identidad

Para todo $x$,

$$
(1f)(x)
=
1\cdot f(x)
=
f(x).
$$

Luego,

$$
1f=f.
$$

---

Como todos los axiomas se verifican, concluimos que

$$
\boxed{C([a,b])\text{ es un espacio vectorial sobre }\mathbb R.}
$$

### El conjunto de matrices $M_{m\times n}(\mathbb{R})$ es un espacio vectorial

Sea

$$
M_{m\times n}(\mathbb{R})
=
\left\{
A=(a_{ij})
:
a_{ij}\in\mathbb R,\;
1\le i\le m,\;
1\le j\le n
\right\}.
$$

Definimos la suma de matrices por

$$
(A+B)_{ij}
=
a_{ij}+b_{ij},
$$

y el producto por un escalar por

$$
(\lambda A)_{ij}
=
\lambda a_{ij},
\qquad
\lambda\in\mathbb R.
$$

Sean

$$
A=(a_{ij}),\qquad
B=(b_{ij}),\qquad
C=(c_{ij}),
$$

y sean

$$
\alpha,\beta\in\mathbb R.
$$

Verificaremos los axiomas de espacio vectorial.

---

#### 1. Clausura de la suma

Como

$$
a_{ij},b_{ij}\in\mathbb R,
$$

y los números reales son cerrados bajo la suma,

$$
a_{ij}+b_{ij}\in\mathbb R
$$

para todo $i$ y $j$.

Por lo tanto,

$$
A+B\in M_{m\times n}(\mathbb R).
$$

---

#### 2. Conmutatividad de la suma

Para toda entrada de la matriz,

$$
a_{ij}+b_{ij}
=
b_{ij}+a_{ij},
$$

ya que la suma en $\mathbb R$ es conmutativa.

Por consiguiente,

$$
A+B=B+A.
$$

---

#### 3. Asociatividad de la suma

Para toda entrada,

$$
(a_{ij}+b_{ij})+c_{ij}
=
a_{ij}+(b_{ij}+c_{ij}),
$$

porque la suma en $\mathbb R$ es asociativa.

Luego,

$$
(A+B)+C
=
A+(B+C).
$$

---

#### 4. Existencia del elemento neutro

Sea

$$
O=
(0)_{m\times n},
$$

es decir, la matriz cuyos elementos son todos cero.

Entonces,

$$
A+O=A.
$$

---

#### 5. Existencia del inverso aditivo

Definimos

$$
-A=(-a_{ij}).
$$

Entonces,

$$
A+(-A)=O.
$$

---

#### 6. Clausura del producto por escalares

Como

$$
\alpha,a_{ij}\in\mathbb R,
$$

se tiene

$$
\alpha a_{ij}\in\mathbb R.
$$

Por lo tanto,

$$
\alpha A
\in
M_{m\times n}(\mathbb R).
$$

---

#### 7. Distributividad respecto de la suma de matrices

Para cada entrada,

$$
\begin{aligned}
(\alpha(A+B))_{ij}
&=
\alpha(a_{ij}+b_{ij})\\
&=
\alpha a_{ij}+\alpha b_{ij}\\
&=
(\alpha A+\alpha B)_{ij}.
\end{aligned}
$$

Luego,

$$
\alpha(A+B)
=
\alpha A+\alpha B.
$$

---

#### 8. Distributividad respecto de la suma de escalares

Para cada entrada,

$$
\begin{aligned}
((\alpha+\beta)A)_{ij}
&=
(\alpha+\beta)a_{ij}\\
&=
\alpha a_{ij}+\beta a_{ij}\\
&=
(\alpha A+\beta A)_{ij}.
\end{aligned}
$$

Por lo tanto,

$$
(\alpha+\beta)A
=
\alpha A+\beta A.
$$

---

#### 9. Compatibilidad del producto por escalares

Para cada entrada,

$$
\begin{aligned}
(\alpha(\beta A))_{ij}
&=
\alpha(\beta a_{ij})\\
&=
(\alpha\beta)a_{ij}\\
&=
((\alpha\beta)A)_{ij}.
\end{aligned}
$$

Así,

$$
\alpha(\beta A)
=
(\alpha\beta)A.
$$

---

#### 10. Existencia del elemento identidad

Para cada entrada,

$$
(1A)_{ij}
=
1\cdot a_{ij}
=
a_{ij}.
$$

Luego,

$$
1A=A.
$$

---

Como todos los axiomas de espacio vectorial se verifican, concluimos que

$$
\boxed{
M_{m\times n}(\mathbb R)
\text{ es un espacio vectorial sobre }
\mathbb R.
}
$$
