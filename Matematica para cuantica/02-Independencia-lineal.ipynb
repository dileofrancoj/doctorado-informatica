# Dependencia e independencia lineal

Hasta ahora sabemos qué es un espacio vectorial y que sus elementos reciben el nombre de vectores. Sin embargo, surge una pregunta natural:

> **¿Cuándo un conjunto de vectores aporta información nueva y cuándo alguno de ellos puede obtenerse a partir de los demás?**

La respuesta a esta pregunta nos lleva al concepto de **dependencia lineal**.

## Definición

Sea $V$ un espacio vectorial y sea

$$
\{v_1,v_2,\ldots,v_n\}\subseteq V.
$$

Decimos que los vectores son **linealmente dependientes** si existen escalares

$$
\alpha_1,\alpha_2,\ldots,\alpha_n\in\mathbb R,
$$

no todos iguales a cero, tales que

$$
\boxed{
\alpha_1v_1+\alpha_2v_2+\cdots+\alpha_nv_n=0.
}
$$

Es decir, existe una combinación lineal no trivial que produce el vector nulo.

---

## Independencia lineal

Decimos que los vectores

$$
\{v_1,v_2,\ldots,v_n\}
$$

son **linealmente independientes** si la única solución de

$$
\alpha_1v_1+\alpha_2v_2+\cdots+\alpha_nv_n=0
$$

es

$$
\boxed{
\alpha_1=\alpha_2=\cdots=\alpha_n=0.
}
$$

En otras palabras, ninguno de los vectores puede escribirse como combinación lineal de los restantes.

---

## Interpretación geométrica

La independencia lineal significa que **cada vector aporta una nueva dirección** dentro del espacio vectorial.

Por el contrario, si un vector puede obtenerse combinando los demás, entonces no agrega ninguna dirección nueva y el conjunto es linealmente dependiente.

Este concepto será fundamental para definir posteriormente las nociones de **base** y **dimensión** de un espacio vectorial.

# Sistema generador

Sea $V$ un espacio vectorial y sea $S=\{v_1,v_2,\ldots,v_n\}\subseteq V$ ,decimos que $S$ es un **sistema generador** de $V$ si todo vector del espacio puede escribirse como una combinación lineal de los vectores de $S$.

Es decir, para todo $v\in V,$ existen escalares $
\alpha_1,\alpha_2,\ldots,\alpha_n
$$

tales que

$$
\boxed{
v=\alpha_1v_1+\alpha_2v_2+\cdots+\alpha_nv_n.
}
$$

En este caso decimos que

$$
\operatorname{span}(S)=V,
$$

o equivalentemente,

$$
\langle v_1,v_2,\ldots,v_n\rangle=V.
$$

---

## Interpretación geométrica

Los vectores del sistema generador son suficientes para construir cualquier otro vector del espacio mediante combinaciones lineales.

Por ejemplo,

- un único vector genera una recta;
- dos vectores linealmente independientes generan un plano;
- tres vectores linealmente independientes generan todo $\mathbb R^3$.

# Base de un espacio vectorial

Sea

$$
V
$$

un espacio vectorial.

Un conjunto

$$
B=\{v_1,v_2,\ldots,v_n\}
$$

es una **base** de $V$ si satisface simultáneamente las siguientes propiedades:

1. Los vectores son linealmente independientes.

2. Los vectores generan todo el espacio.

Es decir,

$$
\boxed{
\begin{aligned}
&v_1,v_2,\ldots,v_n
\text{ son linealmente independientes},\\
\\
&\operatorname{span}(B)=V.
\end{aligned}
}
$$

---

## Interpretación geométrica

Una base representa el conjunto **mínimo de vectores necesarios** para describir completamente un espacio vectorial.

- Si quitamos cualquiera de sus vectores, deja de generar todo el espacio.

- Si agregamos un nuevo vector que pueda escribirse como combinación lineal de los anteriores, aparece redundancia y deja de ser linealmente independiente.

En consecuencia, una base contiene exactamente la información necesaria para describir todos los vectores del espacio, sin información repetida.

---

## Ejemplos

### Base canónica de $\mathbb R^2$

$$
\boxed{
\left\{
(1,0),
(0,1)
\right\}
}
$$

Todo vector

$$
(x,y)
$$

puede escribirse como

$$
(x,y)
=
x(1,0)+y(0,1).
$$

---

### Base canónica de $\mathbb R^3$

$$
\boxed{
\left\{
(1,0,0),
(0,1,0),
(0,0,1)
\right\}
}
$$

Todo vector

$$
(x,y,z)
$$

puede escribirse como

$$
(x,y,z)
=
x(1,0,0)
+
y(0,1,0)
+
z(0,0,1).
$$

---

## Idea fundamental

Una base es un conjunto de vectores que

- genera todo el espacio;
- no posee redundancias.

Es decir,

$$
\boxed{
\text{Base}
=
\text{Sistema generador}
+
\text{Independencia lineal}.
}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

v1 = np.array([2,1])
v2 = np.array([4,2])

fig, ax = plt.subplots(figsize=(7,7))

t = np.linspace(-3,3,100)
ax.plot(t*v1[0], t*v1[1], '--', linewidth=2, label='Recta generada')

ax.quiver(0,0,v1[0],v1[1],
          angles='xy',
          scale_units='xy',
          scale=1,
          color='royalblue',
          width=0.008,
          label='$v_1$')

ax.quiver(0,0,v2[0],v2[1],
          angles='xy',
          scale_units='xy',
          scale=1,
          color='crimson',
          width=0.008,
          label='$v_2=2v_1$')

ax.set_xlim(-7,7)
ax.set_ylim(-7,7)

ax.set_aspect('equal')
ax.grid(True)
ax.axhline(0,color='black')
ax.axvline(0,color='black')

plt.title("Vectores linealmente dependientes")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

v1 = np.array([2,1])
v2 = np.array([-1,2])

fig, ax = plt.subplots(figsize=(7,7))


t = np.linspace(-3,3,100)
ax.plot(t*v1[0],t*v1[1],'--')


ax.plot(t*v2[0],t*v2[1],'--')


ax.quiver(0,0,v1[0],v1[1],
          angles='xy',
          scale_units='xy',
          scale=1,
          color='royalblue',
          width=0.008,
          label='$v_1$')

ax.quiver(0,0,v2[0],v2[1],
          angles='xy',
          scale_units='xy',
          scale=1,
          color='crimson',
          width=0.008,
          label='$v_2$')


P = np.array([
    [0,0],
    v1,
    v1+v2,
    v2,
    [0,0]
])

ax.plot(P[:,0],P[:,1],linewidth=2)

ax.fill(P[:,0],P[:,1],alpha=0.2)

ax.set_xlim(-5,5)
ax.set_ylim(-5,5)

ax.axhline(0,color='black')
ax.axvline(0,color='black')

ax.grid(True)
ax.set_aspect('equal')

plt.title("Vectores linealmente independientes")
plt.legend()

plt.show()
```
