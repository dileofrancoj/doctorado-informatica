# Números complejos y mecánica cuántica.

Por muchas razones, el concepto de número tuvo que extenderse más allá de la recta real. La historia detrás de esta extensión es fascinante y está llena de resistencia, intuiciones brillantes y descubrimientos inesperados. (Acá podés insertar la historia de Cardano, Bombelli y el desarrollo de los números complejos).

Uno de los primeros problemas que motivó esta extensión fue la resolución de ciertas ecuaciones algebraicas. Consideremos, por ejemplo, la ecuación $x^2 = -1.$

Esta ecuación no tiene solución en el conjunto de los números reales, ya que el cuadrado de cualquier número real es siempre mayor o igual que cero.

Frente a este problema tenemos dos alternativas: aceptar que la ecuación simplemente no tiene solución o preguntarnos si es posible ampliar nuestro sistema numérico para que sí la tenga. Históricamente, este fue el mismo tipo de razonamiento que llevó a extender los números racionales hacia los irracionales. (Los pitagóricos, por ejemplo, se encontraron con un obstáculo similar al intentar resolver la ecuación $x^2 = 2$).

La idea consiste en incorporar un nuevo objeto que satisfaga precisamente aquello que los números reales no pueden hacer. Así nace el símbolo (i), definido por la propiedad $i^2 = -1$.

Naturalmente, introducir un nuevo símbolo no alcanza (si no cualquiera haría mates, ¿no?). Si queremos que forme parte de un sistema numérico útil, debemos poder sumarlo, restarlo y multiplicarlo igual que hacemos con los números reales. Esto nos obliga a considerar no solo el número ($i$), sino también expresiones como: $2i,-3i, \frac{1}{2}i$ y en general, todo número de la forma $a+bi$ (donde $a, b \in \mathbb{R}$).

## Intuición

Hasta acá todo bien... pero muchos estarán pensando: **¿dónde vive todo esto?** Al fin y al cabo, los números deberían representar "cosas", ¿no?

Bueno... quizá arrastremos esa idea desde la primaria, pero no necesariamente tiene que ser así.

El número (i) no representa una cantidad mayor o menor que cero, ni puede compararse con los números reales. Es un objeto completamente nuevo que nos obliga a abandonar la intuición —tan natural como limitada— de que **todo número debe vivir sobre una recta**.

El game changer es comprender que un número no necesita representar una cantidad para ser útil. Lo único realmente importante es que las reglas con las que operamos sean coherentes y consistentes. Si esas reglas permiten construir una teoría sin contradicciones, entonces esos nuevos objetos merecen ser estudiados.

Algo parecido ocurre con los vectores. En la escuela solemos aprender que un vector es una flecha, pero en matemática un vector también puede ser una función, una matriz o incluso un polinomio. La intuición inicial es útil para empezar, pero no alcanza para entender toda la teoría.

## Operaciones

Si queremos que el símbolo $i$ forme parte de un sistema numérico útil, no alcanza con definir $i^2 = -1$ y listo el pollo. También debemos poder hacer con él las "mismas" operaciones que hacemos con $\mathbb{R}$. En principio, todo número complejo puede escribirse como $z = a + bi$ (como ya mencionamos)

### Suma

Para determinar la operación binaria de suma entre dos complejos $z_1 = a_1 + b_1i$ y $z_2 = a_2 + b_2i$ se suma la parte real de cada uno y la parte imaginaria: $z_3 = z_1 + z_2 = (a_1 + a_2) + (b_1+b_2)i$

### Multiplicacion
Se multiplican exactamente igual que dos binomios, recordando únicamente que $i^2 = -1$. Entonces:

$(a+bi)(c+di)=ac+adi+bci+(bd)i^2$ y reemplazando:

$=(ac-bd)+(ad+bc)i$

### Conjugado
A cada número complejo $z = a + bi$ le asociamos otro número llamado conjugado, definido por $\overline z=a-bi$. Geométricamente, el conjugado refleja al número respecto del eje real. Surge una propiedad muy útil al multiplicarlos:

$z\overline z=(a+bi)(a-bi)=a^2+b^2$.

Es decir un número complejo por su conjugado siempre da un número real *no negativo*. 

> Esta simple observación será la clave para definir el módulo de un número complejo y, más adelante, la norma de un vector en espacios de Hilbert.



### Interpretación geométrica

Durante bastante tiempo los números complejos fueron vistos como una simple curiosidad algebraica: una herramienta útil para resolver ecuaciones, pero sin una interpretación clara. Recién a comienzos del siglo XIX, gracias a los trabajos de Argand y Gauss, apareció una representación geométrica que terminó de darles sentido.

La idea es sorprendentemente sencilla. En lugar de pensar al número complejo $z = a + bi$ como una expresión algebraica, lo representamos como el punto ((a,b)) del plano. La parte real determina la coordenada horizontal y la parte imaginaria la vertical.



Esta representación nos permite hablar de distancias y ángulos exactamente igual que en el plano cartesiano.

Por ejemplo, la distancia del punto (z) al origen se obtiene aplicando el teorema de Pitágoras:

$|z|=\sqrt{a^2+b^2}$

A esta cantidad se la denomina módulo del número complejo. Si el número complejo tiene únicamente parte real (es decir, (b=0)), el módulo coincide con el valor absoluto que conocemos desde siempre.

Si ya estás familiarizado con el álgebra lineal, esta expresión tiene una interpretación todavía más elegante. El plano complejo puede verse como un espacio vectorial con un producto interno natural, definido por

$\langle z,w\rangle = z\overline{w}$

La norma inducida por ese producto interno satisface

$
|z|^2=\langle z,z\rangle
=z\overline{z}
=a^2+b^2,$

de donde obtenemos nuevamente $|z|=\sqrt{a^2+b^2}.$

Es decir, el módulo no es otra cosa que la longitud del vector que representa al número complejo.

### Módulo 1.
Un conjunto particularmente importante es el de los números complejos cuyo módulo vale uno. Todos ellos satisfacen

$|z|=1$,

o, de manera equivalente, $a^2+b^2=1.$ Geométricamente, estos puntos forman el círculo unitario, es decir, la circunferencia de radio uno centrada en el origen.



## Forma trigonométrica

Hasta ahora describimos un número complejo mediante sus coordenadas cartesianas: $z = a +bi$. Sin embargo, existe otra forma -igual de válida- que muchas veces es más conveniente para representar al número $z$. Podemos cambiar la representación cartesiana a otro sistema de coordenadas. Nos iremos a polares. Las coordenadas polares (al estar en el plano) se representan justamente la distancia al origen y el ángulo que forma el segmento con el eje real positivo.

Como ya adelantamos, la distancia viene dada por: $\rho=|z|=\sqrt{a^2+b^2}$. Solo nos falta introducir el ángulo, que normalmente se representa mediante la letra griega $\theta$ y recibe el nombre de argumento del número complejo.

Geométricamente, $\theta$ es el ángulo medido desde el eje real positivo hasta el vector que une el origen con el punto z.

Si observamos el triángulo rectángulo formado por el punto, aparecen inmediatamente las razones trigonométricas:

$
\cos\theta=\frac{a}{\rho},
\qquad
\sin\theta=\frac{b}{\rho}.$

Despejando las coordenadas obtenemos

$
a=\rho\cos\theta,
\qquad
b=\rho\sin\theta.
$

Ahora simplemente reemplazamos estas expresiones en la forma habitual del número complejo:

$
z=a+bi.
$

Entonces,
$
z=\rho\cos\theta+\rho\sin\theta,i,
$

y factorizando el módulo,

$
\boxed{z=\rho(\cos\theta+i\sin\theta).}
$

Esta expresión recibe el nombre de forma trigonométrica (o forma polar) de un número complejo.

Es importante notar que este no es un número nuevo. Seguimos hablando exactamente del mismo punto del plano; simplemente lo estamos describiendo con otro sistema de coordenadas.

> Cada representación resulta más conveniente según el problema que queramos resolver.
Por ejemplo, para sumar números complejos suele ser más cómodo trabajar con la forma cartesiana. En cambio, para multiplicarlos, dividirlos o calcular potencias y raíces, la forma trigonométrica revela una estructura sorprendentemente simple que veremos en la siguiente sección.

## ¿Por qué aparecen complejos en la mecánica cuántica?

Lo curioso es que estos números, que durante siglos parecieron un artificio algebraico, terminaron siendo el lenguaje natural para describir ondas, electricidad, procesamiento de señales y, sobre todo, la mecánica cuántica. El famoso símbolo que presentamos $i$ que nació -sencillamente- para resolver una ecuación aparece hoy en la ecuación de Schrödinger y en prácticamente toda la teoría cuántica. *Pero... ¿por qué?*
